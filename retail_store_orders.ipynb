```python
# the downloaded file is a zip file, we gotto extract th edata from it using the zip library 
import zipfile
zip_ref=zipfile.ZipFile("Orders.csv.zip")
zip_ref.extractall() # extracts the desired files from the zip files
zip_ref.close()  # closes the file
```

```python
## now that the data has been uploaded and unzipped lets do some eda
import pandas as pd
import numpy as np
df= pd.read_csv("Orders.csv",na_values=['Not Available', 'unknown'])
```

```python
df.head(20)
```

```python
# we can obsereve that from shipment there are some unknown values , lets look into its distinct values

df["Ship Mode"].unique()
```

```python
## the column names have spaces so we got to fix them 
df
```

```python
## we can use replace but it is time consuming , so we wll extract the columns and convert it into string 
df.columns=df.columns.str.lower()
df
```

```python
## all of the column names are converted into lower case, non insteadd of space lets insert underscore
df.columns = df.columns.str.replace(" ","_")
df.head(5)
```

```python
## if you observe there is cp , discount percentage but no discount price so lets create our own column discount price
df["discount"]=df['list_price']*df['discount_percent']*0.01
```

```python
df['discount']
```

```python
df.head(5)
```

```python
## so theres cp,llist price, discount , lets draw a sp by subtracting discount from list price
df['sale_price']=df['list_price']-df['discount']
```

```python
df.head(4)
```

```python
# profit sp-cp
df['profit']= df['sale_price']-df['cost_price']
```

```python
df['profit']
```

```python
df.dtypes
```

```python
# we see that the order date is as object so we got to convert it into the date datatype using hte pandas dataframe
df['order_date']= pd.to_datetime(df['order_date'],format= "%Y-%m-%d")
```

```python
df.dtypes
```

```python
## now that weve have sp we dont need list price cp and discount percentage for analysis so we'll drop them using the .drop function
df.drop(columns=['list_price','cost_price','discount_percent'],inplace=True)
```
